```python
import tensorflow as tf
import tensorflow.keras as keras
from itertools import groupby
import numpy as np
```

```python
def faiter(file):
    """
    funkcja zwraca generator. kazdy loop generatora 
    zawiera header oraz połączoną z poj. linii sekwencję.
    sekwencja grupowana jest do headera za pomocą groupby().
    """
    fh = open(file)
    for group, group_items_iter in groupby(fh, lambda line: line[0] == ">"):
        header_seq_joined = "".join(group_items_iter)
        yield header_seq_joined

def single_sequence(faiter_output):
    """
    funkcja zwraca pojedyncza sekwencję
    wraz z jej headerem w oddzielnych zmiennych
    """
    header = faiter_output.__next__().replace('\n', '')
    sequence = faiter_output.__next__().replace('\n', '')
    return (header, sequence)

def check_no_of_sequences(file):
    sequences_count = 0
    f = open(file)

    for line in f:
        if line.startswith(">"):
            sequences_count += 1
    return sequences_count
```

```python
# .npy files generated, now load files and combine them (after restarting kernel)

PCT = np.load('lncrna_rep/PCT_300nt.npy')
NCT = np.load('lncrna_rep/NCT_300nt.npy')
```

```python
PCT_train = PCT[:34000]
PCT_test = PCT[34000:]

NCT_train = NCT[:34000]
NCT_test = NCT[34000:]
PCT_train.shape
```

```python
NCT_test.shape
```

```python
x_train = np.vstack([PCT_train, NCT_train])
x_train.shape

x_test = np.vstack([PCT_test, NCT_test])
x_train.shape
```

```python
y_train_1 = np.ones(34000)
y_train_2 = np.zeros(34000)
y_train = np.hstack([y_train_1, y_train_2])
y_train = keras.utils.to_categorical(y_train, 2)
y_train.shape

y_test_1 = np.ones(9877) # 1 = PC, 0 = NC (?)
y_test_2 = np.zeros(5244)
y_test = np.hstack([y_test_1, y_test_2])
y_test = keras.utils.to_categorical(y_test, 2)
y_test.shape
```

```python
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Dropout, Flatten, Reshape, GlobalAveragePooling1D, Activation
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Conv1D, MaxPooling1D
#from tensorflow.keras.utils import np_utils
```

```python
model_m = Sequential()

model_m.add(tf.keras.layers.Masking(mask_value=0.,input_shape=(3000, 4)))


model_m.add(Conv1D(50, 15, activation='relu', ))

model_m.add(Conv1D(50, 15, activation='relu'))
model_m.add(MaxPooling1D(3))
model_m.add(Conv1D(80, 15, activation='relu'))
model_m.add(Conv1D(80, 15, activation='relu'))
model_m.add(GlobalAveragePooling1D())
model_m.add(Dropout(0.5))
model_m.add(Dense(2, activation='softmax'))
print(model_m.summary())
```

```python
#callbacks_list = [
#    keras.callbacks.ModelCheckpoint(
 #       filepath='best_model.{epoch:02d}-{val_loss:.2f}.h5',
  #      monitor='val_loss', save_best_only=True),
   # )
#]

callbacks_list = [keras.callbacks.EarlyStopping(monitor='acc', patience=1)]

model_m.compile(loss='categorical_crossentropy',
                optimizer='adam', metrics=['accuracy'])

BATCH_SIZE = 400
EPOCHS = 10

history = model_m.fit(x_train,
                      y_train,
                      batch_size=BATCH_SIZE,
                      epochs=EPOCHS,
                      callbacks=callbacks_list)
                      
```
