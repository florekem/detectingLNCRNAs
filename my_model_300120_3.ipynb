```python
import numpy as np
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Dropout, GlobalMaxPool1D, GlobalAveragePooling1D, Activation #Embedding
from tensorflow.keras.layers import Conv1D, MaxPooling1D, AveragePooling1D, Flatten
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import SGD
```

```python
x_train = np.load('../lncrna_rep/3000maxseq_len_padded_w0_short/x_train.npy')
y_train = np.load('../lncrna_rep/3000maxseq_len_padded_w0_short/y_train.npy')
x_test = np.load('../lncrna_rep/3000maxseq_len_padded_w0_short/x_test.npy')
y_test = np.load('../lncrna_rep/3000maxseq_len_padded_w0_short/y_test.npy')
```

### my_model_300120_3

```python
model = Sequential()
model.add(Conv1D(512, 3, activation='relu', input_shape=(3000, 4)))

model.add(AveragePooling1D(3))
model.add(Flatten()) # chyba powinno byc odwrotnie z dropout, ale nie spr roznicy
model.add(Dropout(0.9))
model.add(Dense(2, activation='sigmoid'))


print(model.summary())


```

```python
callbacks_list = [EarlyStopping(monitor='acc', patience=5)]

model.compile(loss='binary_crossentropy',
              optimizer='rmsprop',
              metrics=['accuracy'])

BATCH_SIZE = 128
EPOCHS = 10

history = model.fit(x_train,
                      y_train,
                      batch_size=BATCH_SIZE,
                      epochs=EPOCHS,
                      callbacks=callbacks_list,
                      validation_data=(x_test, y_test))

```

```python
model.save('../lncrna_rep/my_model_300120_3.h5')
```

### my_model_300120_4 (avarage zamiast flatten, adam)

```python
model = Sequential()
model.add(Conv1D(512, 3, activation='relu', input_shape=(3000, 4)))

model.add(AveragePooling1D(3))
model.add(Dropout(0.9))
model.add(GlobalAveragePooling1D())
model.add(Dense(2, activation='sigmoid'))


print(model.summary())
```

```python
callbacks_list = [EarlyStopping(monitor='acc', patience=5)]

model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

BATCH_SIZE = 128
EPOCHS = 10

history = model.fit(x_train,
                      y_train,
                      batch_size=BATCH_SIZE,
                      epochs=EPOCHS,
                      callbacks=callbacks_list,
                      validation_data=(x_test, y_test))
```
